# 03 · Surveillance policies: echoes saved vs detection delay (Idea 2)

Reproduce: `cd notebooks/analysis && /data/abar/alexenv/bin/python run_all.py policy` (13 policy runs in parallel, ~2 min).

The simulator is re-run with a **sequential visit generator**: at every echo the policy decides the next echo time from the online features (the same function that built the training table). Symptom-triggered echoes still happen, a positive single echo triggers a confirmation echo at 3–6 months (all policies except the pure guideline), reintervention depends on *detected* disease, and every policy sees the same pre-drawn noise and the same latent trajectories (common random numbers).

Policies: guideline (annual), guideline + confirmation echo, de-escalation (biennial after 5 y if nothing seen), time-since-implant-only (null: same interval rule, no echo information), risk-adapted (largest interval in {6, 12, 18, 24, 36} months with predicted confirmed-SVD risk ≤ τ and stage-3 risk ≤ 2%), oracle (echo at the latent crossing; the ceiling).

Pass criterion (docs/03 §0): ≥25% fewer echoes per patient-decade than the guideline with ≤3 months extra mean and P90 detection delay.
